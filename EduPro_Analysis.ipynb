#### Learning Depth Index

```python
advanced_enrollments = (
    level_enrollment[level_enrollment["CourseLevel"] == "Advanced"]
    .groupby("UserID")
    .size()
    .reset_index(name="AdvancedEnrollments")
)

print(advanced_enrollments.head(10))
```

```python
learner_profile = learner_profile.merge(
    advanced_enrollments,
    on="UserID",
    how="left"
)

learner_profile["AdvancedEnrollments"] = (
    learner_profile["AdvancedEnrollments"].fillna(0)
)

print(learner_profile[["UserID", "EnrollmentFrequency", "AdvancedEnrollments"]].head(10))
```

```python
learner_profile["LearningDepthIndex"] = (
    learner_profile["AdvancedEnrollments"]
    / learner_profile["EnrollmentFrequency"]
)

print(
    learner_profile[
        ["UserID", "AdvancedEnrollments", "EnrollmentFrequency", "LearningDepthIndex"]
    ].head(10)
)
```

#### Separate Numerical and Categorical Features

```python
numerical_features = [
    "Age",
    "TotalCoursesEnrolled",
    "EnrollmentFrequency",
    "AverageSpending",
    "AverageCourseRating",
    "CategoryDiversity",
    "AdvancedEnrollments",
    "LearningDepthIndex"
]

categorical_features = [
    "Gender",
    "PreferredCategory",
    "PreferredLevel"
]

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)
```

#### Encode the Categorical Features

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
encoded_categorical = encoder.fit_transform(
    learner_profile[categorical_features]
)

print(encoded_categorical.shape)
```

#### Combine Numerical + Encoded Features

```python
import pandas as pd

numerical_data = learner_profile[numerical_features].copy()

encoded_data = pd.DataFrame(
    encoded_categorical,
    columns=encoder.get_feature_names_out(categorical_features),
    index=learner_profile.index
)

print("Numerical data shape:", numerical_data.shape)
print("Encoded data shape:", encoded_data.shape)
```

```python
clustering_data = pd.concat(
    [numerical_data, encoded_data],
    axis=1
)

print("Final clustering data shape:", clustering_data.shape)
print(clustering_data.head())
```

#### Scale the Features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
scaled_data = scaler.fit_transform(clustering_data)

print("Scaled data shape:", scaled_data.shape)
print(scaled_data[:5])
```

# Section 8 — Student Segmentation using K-Means.

#### Test Different Numbers of Clusters

```python
from sklearn.cluster import KMeans

inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(scaled_data)
    inertia.append(kmeans.inertia_)

print(inertia)
```

#### Plot the Elbow Curve

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    inertia,
    marker="o"
)

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.xticks(range(2, 11))
plt.grid(True)

plt.tight_layout()
plt.show()
```

#### Calculate Silhouette Scores

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(scaled_data)
    
    score = silhouette_score(
        scaled_data,
        labels
    )
    
    silhouette_scores.append(score)

print(silhouette_scores)
```

#### Minimum Rating Filter

```python
def recommend_courses(user_id, top_n=5):

    learner = learner_profile[
        learner_profile["UserID"] == user_id
    ].iloc[0]

    preferred_category = learner["PreferredCategory"]
    preferred_level = learner["PreferredLevel"]

    taken_courses = existing_courses.get(user_id, set())

    recommendations = course_recommendation_data[
        (~course_recommendation_data["CourseID"].isin(taken_courses))
        & (course_recommendation_data["CourseRating"] >= 3.0)
    ].copy()

    recommendations["Score"] = (
        (recommendations["CourseCategory"] == preferred_category).astype(int) * 5
        + (recommendations["CourseLevel"] == preferred_level).astype(int) * 6
        + recommendations["CourseRating"] * 2
        + (
            recommendations["EnrollmentCount"]
            / recommendations["EnrollmentCount"].max()
        )
    )

    recommendations = recommendations.sort_values(
        "Score",
        ascending=False
    )

    return recommendations.head(top_n)
```
